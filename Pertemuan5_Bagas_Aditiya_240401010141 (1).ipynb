```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
df = sns.load_dataset("titanic")
df
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='class')
plt.title('Jumlah Penumpang Berdasarkan Kelas')
plt.show()
```

```python
df['class'].value_counts()
```

Berdasarkan data pada grafik Bar chart tersebut kelas 3 memiliki jumlah penumpang terbanyak yaitu 491 orang, sedangkan kelas 2 memiliki jumlah paling sedikit yaitu 184 orang.

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='age', bins=20, kde=True)
plt.title('Distribusi Usia Penumpang')
plt.show()
```

```python
df['age'].describe()
```

Berdasarkan grafik histogram tersebut rata-rata usia penumpang adalah sekitar 29,7 tahun dengan usia termuda 0,42 tahun dan tertua 80 tahun

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='class', y='fare')
plt.title('Sebaran Tarif Berdasarkan Kelas')
plt.show()
```

```python
df.groupby('class')['fare'].describe()
```

Berdasarkan grafik box plot tersebut rata-rata tarif kelas 1 jauh lebih tinggi dibandingkan kelas 2 dan kelas 3.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='age',
    y='fare',
    hue='survived'
)
plt.title('Hubungan Usia dan Tarif')
plt.show()
```

```python
df[['age','fare']].corr(numeric_only=True)
```

Berdasarkan grafik scatter plot tersebut hubungan antara usia dan tarif sangat lemah karena nilai korelasinya hanya sekitar 0,09
